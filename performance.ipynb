```python
import pandas as pd
import numpy as np
```

```python
bart = pd.read_csv('bart.csv')
t5 = pd.read_csv('t5.csv')
pegasus = pd.read_csv('pegasus.csv')
```

```python
df = pd.read_csv('bart.csv')
```

```python
bart.columns = ['Summary', 'Text', 'bart']
t5.columns = ['Summary', 'Text', 't5']
pegasus.columns = ['Summary', 'Text', 'pegasus']
```

```python
df = df[['Text', 'Summary']]
```

```python
df['bart'] = bart['bart']
df['t5'] = t5['t5']
df['pegasus'] = pegasus['pegasus']
```

```python
from nltk.translate.bleu_score import sentence_bleu
from tqdm import tqdm
from datasets import load_dataset, load_metric
```

```python
df
```

```python
for i, row in tqdm(df[:].iterrows()):
    bart = sentence_bleu([row['Summary'].split()], row['bart'].split())
    t5 = sentence_bleu([row['Summary'].split()], row['t5'].split())
    pegasus = sentence_bleu([row['Summary'].split()], row['pegasus'].split())
    if((bart>=t5) & (bart>=pegasus)):
        df.at[i, 'final'] = row['bart']
    elif((t5>=bart) & (t5>=pegasus)):
        df.at[i, 'final'] = row['t5']
    elif((pegasus>=bart) & (pegasus>=t5)):
        df.at[i, 'final'] = row['pegasus']
```

```python
metric = load_metric("rouge")
```

```python
df
```

```python
results = metric.compute(predictions=df['final'],
                         references =df['Summary'])
print(results)
```

```python
for key in ["rouge1", "rouge2", "rougeL"]:
    print(key, "low:", results[key].low.fmeasure)
    print(key, "mid:", results[key].mid.fmeasure)
    print(key, "high:", results[key].high.fmeasure, end="\n\n")
```

```python
bleu = []
for i, row in tqdm(df.iterrows()):
  bleu.append(sentence_bleu([row['Summary'].split()], row['final'].split()))
```

```python
print('bleu   :',   np.mean(bleu))
```
