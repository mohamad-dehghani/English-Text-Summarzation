```python
import pandas as pd
import numpy as np
```

```python
add1 = 'bart.csv'
add2 = 't5.csv'
add3 = 'pegasus.csv'
```

```python
df1 = pd.read_csv(add1)
df2 = pd.read_csv(add2)
df3 = pd.read_csv(add3)
```

```python
df = pd.read_csv(add1)
```

```python
df1.columns = ['Summary', 'Text', add1[:-4]]
df2.columns = ['Summary', 'Text', add2[:-4]]
df3.columns = ['Summary', 'Text', add3[:-4]]
```

```python
df = df[['Text', 'Summary']]
```

```python
df[add1[:-4]] = df1[add1[:-4]]
df[add2[:-4]] = df2[add2[:-4]]
df[add3[:-4]] = df3[add3[:-4]]
```

```python
from nltk.translate.bleu_score import sentence_bleu
from tqdm import tqdm
from datasets import load_dataset, load_metric
```

```python
weights1 = (1, 0, 0, 0)
weights2 = (0, 1, 0, 0)
weights3 = (0, 0, 1, 0)
weights4 = (0, 0, 0, 1)
```

```python
for i, row in tqdm(df[:].iterrows()):
    
    bart_score = 0
    t5_socre = 0
    pegasus_score = 0
    
    bart1 = sentence_bleu([row['Text'].split()], row['bart'].split(), weights1)
    t1 = sentence_bleu([row['Text'].split()], row['t5'].split(), weights1)
    pegasus1 = sentence_bleu([row['Text'].split()], row['pegasus'].split(), weights1)
    
    if((bart1>t1)&(bart1>pegasus1)):
        bart_score+=1
        
    elif((t1>bart1)&(t1>pegasus1)):
        t5_socre+=1
        
    elif((pegasus1>bart1)&(pegasus1>t1)):
        pegasus_score+=1
        
    bart2 = sentence_bleu([row['Text'].split()], row['bart'].split(), weights2)
    t2 = sentence_bleu([row['Text'].split()], row['t5'].split(), weights2)
    pegasus2 = sentence_bleu([row['Text'].split()], row['pegasus'].split(), weights2)
    
    if((bart2>t2)&(bart2>pegasus2)):
        bart_score+=1
        
    elif((t2>bart2)&(t2>pegasus2)):
        t5_socre+=1
        
    elif((pegasus2>bart2)&(pegasus2>t2)):
        pegasus_score+=1
    
    bart3 = sentence_bleu([row['Text'].split()], row['bart'].split(), weights3)
    t3 = sentence_bleu([row['Text'].split()], row['t5'].split(), weights3)
    pegasus3 = sentence_bleu([row['Text'].split()], row['pegasus'].split(), weights3)
    
    if((bart3>t3)&(bart3>pegasus3)):
        bart_score+=1
        
    elif((t3>bart3)&(t3>pegasus3)):
        t5_socre+=1
        
    elif((pegasus3>bart3)&(pegasus3>t3)):
        pegasus_score+=1
    
    bart4 = sentence_bleu([row['Text'].split()], row['bart'].split(), weights4)
    t4 = sentence_bleu([row['Text'].split()], row['t5'].split(), weights4)
    pegasus4 = sentence_bleu([row['Text'].split()], row['pegasus'].split(), weights4)
    
    if((bart4>t4)&(bart4>pegasus4)):
        bart_score+=1
        
    elif((t4>bart4)&(t4>pegasus4)):
        t5_socre+=1
        
    elif((pegasus4>bart4)&(pegasus4>t4)):
        pegasus_score+=1
        
    if(bart_score>=2):
        print('sent', i+1, ' :', add1[:-4])
        df.at[i, 'final'] = row[add1[:-4]]
    elif(t5_socre>=2):
        print('sent', i+1, ' :', add2[:-4])
        df.at[i, 'final'] = row[add2[:-4]]
    elif(pegasus_score>=2):
        print('sent', i+1, ' :', add3[:-4])
        df.at[i, 'final'] = row[add3[:-4]]
```

```python
metric = load_metric("rouge")
```

```python
results = metric.compute(predictions=df['final'],
                         references =df['Summary'])
```

```python
for key in ["rouge1", "rouge2", "rougeL"]:
    print(key, "low:", results[key].low.fmeasure)
    print(key, "mid:", results[key].mid.fmeasure)
    print(key, "high:", results[key].high.fmeasure, end="\n\n")
```

```python
bleu = []
for i, row in tqdm(df.iterrows()):
  bleu.append(sentence_bleu([row['Summary'].split()], row['final'].split()))
```

```python
print('bleu   :',   np.mean(bleu))
```
